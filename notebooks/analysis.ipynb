## Data preprocesing

```python
df.drop(columns=["id", "Unnamed: 32"], inplace=True, errors="ignore") # unnecessary column
```

```python
df['diagnosis'] = df['diagnosis'].map({
    "B": 0,
    "M": 1
})
```

```python
print(df["diagnosis"].value_counts())
```

```python
X = df.drop("diagnosis", axis=1) #features

y = df["diagnosis"] #target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y

)
```

```python
test_data = X_test.copy()
test_data["diagnosis"] = y_test

test_data.to_csv("../test_data.csv", index=False)

print("test_data.csv saved successfully!")
```

## import and fit models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier

import joblib
```

```python
logistic_model = LogisticRegression(random_state=42)
decision_tree_model = DecisionTreeClassifier(random_state=42)
knn_model = KNeighborsClassifier(n_neighbors=5)
naive_bayes_model = GaussianNB()
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
) 
```

```python
logistic_model.fit(X_train,y_train)
```

```python
decision_tree_model.fit(X_train, y_train)
```

```python
knn_model.fit(X_train, y_train)
```

```python
naive_bayes_model.fit(X_train, y_train)
```

```python
random_forest_model.fit(X_train, y_train)
```

```python
#save the models
joblib.dump(logistic_model, "../models/logistic.pkl")
joblib.dump(decision_tree_model, "../models/decision_tree.pkl")
joblib.dump(knn_model, "../models/knn.pkl")
joblib.dump(naive_bayes_model, "../models/naive_bayes.pkl")
joblib.dump(random_forest_model, "../models/random_forest.pkl")
```

```python
print("All models trained successfully!")

print("\nSaved Models:")
print("- Logistic Regression")
print("- Decision Tree")
print("- KNN")
print("- Naive Bayes")
print("- Random Forest")
print("- Scaler")
```

## Evaluate Models & Prediction


```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)
```

```python
# Logistic Regression
logistic_pred = logistic_model.predict(X_test)
logistic_prob = logistic_model.predict_proba(X_test)[:, 1]

# Decision Tree
decision_tree_pred = decision_tree_model.predict(X_test)
decision_tree_prob = decision_tree_model.predict_proba(X_test)[:, 1]

# KNN
knn_pred = knn_model.predict(X_test)
knn_prob = knn_model.predict_proba(X_test)[:, 1]

# Naive Bayes
naive_bayes_pred = naive_bayes_model.predict(X_test)
naive_bayes_prob = naive_bayes_model.predict_proba(X_test)[:, 1]

# Random Forest
random_forest_pred = random_forest_model.predict(X_test)
random_forest_prob = random_forest_model.predict_proba(X_test)[:, 1]
```

```python
def evaluate_model(y_true, y_pred, y_prob):

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "AUC": roc_auc_score(y_true, y_prob),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1 Score": f1_score(y_true, y_pred),
        "MCC": matthews_corrcoef(y_true, y_pred)
    }
```

```python
results = {
    "Logistic Regression": evaluate_model(y_test, logistic_pred, logistic_prob),
    "Decision Tree": evaluate_model(y_test, decision_tree_pred, decision_tree_prob),
    "KNN": evaluate_model(y_test, knn_pred, knn_prob),
    "Naive Bayes": evaluate_model(y_test, naive_bayes_pred, naive_bayes_prob),
    "Random Forest": evaluate_model(y_test, random_forest_pred, random_forest_prob)
}
```

```python
metrics_df = (
    pd.DataFrame(results)
      .T
      .reset_index()
      .rename(columns={"index": "Model"})
)

metrics_df
```

```python
metrics_df.to_csv("../outputs/metrics.csv", index=False) #saving it in metrics.csv
```

## Confusion matrix & classification report

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import confusion_matrix, classification_report
```

```python
def plot_confusion_matrix(y_true,y_pred,model_name):
    cm= confusion_matrix(y_true,y_pred)
    plt.Figure(figsize=(5,4))
    sns.heatmap(
        cm,
        annot =True,
        fmt ="d",
        cmap="Blues",
        xticklabels=["benign","Malignant"],
        yticklabels=["benign","Malignant"]

    )
    plt.title(f"{model_name}Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig(f"../outputs/{model_name.lower().replace(' ','_')}_cm.png")
    plt.show()
```

```python
plot_confusion_matrix(y_test, logistic_pred, "Logistic Regression")
plot_confusion_matrix(y_test, decision_tree_pred, "Decision Tree")
plot_confusion_matrix(y_test, knn_pred, "KNN")
plot_confusion_matrix(y_test, naive_bayes_pred, "Naive Bayes")
plot_confusion_matrix(y_test, random_forest_pred, "Random Forest")
```

```python
print("========== Logistic Regression ==========")
print(classification_report(y_test, logistic_pred))

print("========== Decision Tree ==========")
print(classification_report(y_test, decision_tree_pred))

print("========== KNN ==========")
print(classification_report(y_test, knn_pred))

print("========== Naive Bayes ==========")
print(classification_report(y_test, naive_bayes_pred))

print("========== Random Forest ==========")
print(classification_report(y_test, random_forest_pred))
```
